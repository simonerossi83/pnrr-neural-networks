# Where Do PNRR Funds Go in Italy?
## 1. Research proposal and hypotheses

**Project:** Use a neural network to help classify PNRR projects and explore funding by sector, implementing organisation, and territory.

**Status:** Proposal only. No model has been trained and no findings are claimed.

## 1.1 The problem

I aim to make the distribution of Italy’s *Piano Nazionale di Ripresa e Resilienza* (PNRR) easier to understand: which activities receive funding, which organisations implement projects, and how resources are distributed across North, Centre, South, and Islands.

The opportunity for a neural network is to classify Italian project titles into meaningful purpose categories, such as digitalisation, infrastructure, education, and health. This could help when existing classifications are missing or too broad. Recorded categories will remain available alongside model predictions.

Funding totals and geographical comparisons will be calculated directly from the data. The network will predict a **project category**, not invent financial amounts or locations. The scope is the projects covered by the available data; complete coverage of all PNRR funds must be checked.

## 1.2. Why it matters

An understandable view of funding can support public transparency, help citizens explore investments in their area, and help researchers identify territorial and sectoral differences worth investigating. Keeping allocated funding separate from recorded payments also clarifies the difference between financing a project and spending money on it.

Differences in funding alone do not establish unfairness, misuse, or policy impact.

## 1.3. Appropriate data

The following files are already in this workspace; their column headers have been inspected, but their quality and coverage have not yet been assessed.

| Local file | Proposed use |
|---|---|
| `data/progetti.csv` | Project titles, identifiers, PNRR financing, measure links, and implementing bodies |
| `data/misure.csv` and `data/temi.csv` | Link measures to themes and define initial classification labels |
| `data/progetti_territori.csv` | Locate projects and derive geographical groups |
| `data/progetti_pagamenti.csv` | Analyse recorded PNRR payments and reporting dates separately from financing |

OpenPNRR publishes these types of datasets and a [field dictionary](https://openpnrr.openpolis.io/metadati/). Its documentation distinguishes `finanziamento_pnrr` from total project financing and `pagamento_pnrr` from total payments. It also defines the project field `descrizione` as the **measure title**, so this field should not be mistaken for an independent project description.

Before analysis, verify join keys, snapshot dates, missing values, and duplicate records. Keep national or multi-region projects in separate groups unless reliable territorial amounts exist, avoiding repeated counting of the same funding. Use North, Centre, South, and Islands; combine South and Islands explicitly when reporting “Mezzogiorno.”

An implementing body is not necessarily the final beneficiary or supplier. Identifying recipient industries would require additional organisation/industry data; project purpose categories are a first approximation of funded activities, not company industry codes.

## 1.4. Proposed neural network

Start with a small **multilayer perceptron (MLP)**: convert each Italian project title into TF–IDF numerical features, pass them through one or two hidden layers, and output probabilities for broad purpose categories.

Create initial labels by linking projects to measures and themes, then mapping themes to one broad category. These inherited labels are provisional: manually review a sample, resolve ambiguous mappings, and reserve an independently reviewed test sample. If reliable labels cannot be created, pause training and improve the annotation process.

Exclude measure names, measure codes, and identifiers from model inputs because they could reveal the target. Compare the network with a majority-class baseline and TF–IDF logistic regression. Neural networks are useful here only if they demonstrate value beyond these simpler approaches.

## 1.5. Research hypothesis and null hypothesis

**Research question:** Can a neural network classify the purpose of previously unseen PNRR projects from their titles more accurately than a simple text classifier?

**H₁ — Alternative hypothesis:** The MLP achieves higher expected macro-F1 than TF–IDF logistic regression on unseen, independently reviewed projects.

**H₀ — Null hypothesis:** The MLP achieves no higher expected macro-F1 than TF–IDF logistic regression on those projects.

Let $Delta = F1_{mathrm{macro,MLP}} - F1_{mathrm{macro,logistic}}$.

$$H_0: Delta leq 0 qquad H_1: Delta > 0$$

Macro-F1 averages classification performance across categories so that large categories do not dominate the score. This hypothesis tests the model’s usefulness; it does not assume that North or South receives more funding.

## 1.6. What success would look like

- **Useful analysis:** Reproducible tables and charts of PNRR financing and payments by purpose, implementing body, and geographical area, with unknown or unclassified amounts visible.
- **Useful model:** A proposed target of test macro-F1 ≥ 0.80 and an improvement of at least 0.03 over logistic regression. These are project goals, not observed results.
- **Credible evaluation:** Keep related measures, repeated projects, and duplicate titles together when splitting training, validation, and test data. Ensure each evaluated class has sufficient independent groups; use validation data for tuning and evaluate the frozen models once on the test set.
- **Hypothesis assessment:** Estimate a paired bootstrap confidence interval for the macro-F1 difference, resampling independent groups. Reject H₀ at the 5% level only if the one-sided 95% lower confidence bound is above zero. Report uncertainty, class-level errors, and performance across territories.

If evidence is insufficient, fail to reject H₀; this does not prove that the models are equivalent. A reliable funding analysis remains valuable even if the neural network does not outperform the baseline.
